```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv()
```

```python
llm = ChatGroq(model="llama-3.3-70b-versatile", streaming=True)
question = " Tell me About Gen AI"
res = llm.stream(question)
```

```python
for chunk in res:
    print(chunk.content, end="")
```
