```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_chroma import Chroma
```

Docs loading


```python
loader = PyPDFLoader("../data/data_science_syllabus.pdf")
texts = loader.load()
```

Chunking


```python
Splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=30)
split_data = Splitter.split_documents(texts)
```

Embedding

```python
embedded = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview")
```

Storing in vector DB

```python
vector = Chroma.from_documents(
    documents=split_data,
    embedding=embedded
)
```

2ND PHASE


```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
```

Query

```python
query = "full course duraton ?"
llm = ChatGroq(
    model="openai/gpt-oss-120b"
)
```

Get Context

```python
def getContext(query:str):
    data = vector.similarity_search(query=query)
    context = ""
    for con in data:
        context += con.page_content
        
    return {
        "context": context,
        "question": query
    }
```

```python
prompt = PromptTemplate.from_template(
    """
    You are a helpful assistant and provide answerd based on the context for user question. and 
    if you don't know the answer, then you can say that 'I dont know.'
    Context: {context}
    Question: {question}
"""
)
```

```python
rag_chain = getContext | prompt | llm
res = rag_chain.invoke(query)
print(res.content)
```
