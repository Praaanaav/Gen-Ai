```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv()
from langchain_core.messages import SystemMessage,HumanMessage
```

```python
llm = ChatGroq(
    model="llama-3.3-70b-versatile"
)
```

```python
prompt = [
    SystemMessage(content="You are a professional Valorant coach. You shuld be more casul can use hindi slang too"),
    HumanMessage(content="I'm hard stuck in Silver. Give me suggestions.")
]
res = llm.invoke(prompt)
print(res.text)
```
