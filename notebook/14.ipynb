```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langchain_core.tools import tool
from langchain.agents import create_agent
load_dotenv()
```

```python
@tool
def add_numbers(a:int, b:int):
    """
        it will return the sum of two numbers

        Args:
            a : Number One
            b : Number Two
    """
    return a + b


@tool
def multiply_numbers(a:int, b:int):
    """
        it will return the product of two numbers

        Args:
            a : Number One
            b : Number Two
    """
    return a * b
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-120b"
)
```

```python
agent = create_agent(
    model=llm,
    tools=[multiply_numbers, add_numbers],
    system_prompt="You are a math teacher, and always use tool for calculation and explain a little."
)
```

```python
res = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "What is 2 + 30 * 10?"
        }
    ]
})

print(res["messages"][-1].content)
```
